```python
from scipy.special import kv, k0
import time
import numpy as np

c = 0.71294
o = np.ones((50,50))
n_iter = int(1e4)

def bessel_function_slow():
    for i in range(n_iter):
        kv(0,c * o)
    return None

def bessel_function_fast():
    for i in range(n_iter):
        k0(c * o)
    return None

def cos_function():
    for i in range(n_iter):
        np.cos(c * o)
    return None

def timeit(fcn):
    start_time = time.perf_counter()
    fcn()
    end_time = time.perf_counter()
    execution_time = end_time - start_time
    return execution_time/n_iter

b_time_slow = timeit(bessel_function_slow) * 1e6
b_time_fast = timeit(bessel_function_fast) * 1e6
c_time = timeit(cos_function) * 1e6

# matlab command: o=ones(50); fb=@()besselj(1,0.46737*o); fc=@()cos(0.46737*o); [timeit(fb) timeit(fc)]
mat_b_time = 7.4e-5 * 1e6
mat_c_time = 1.1e-5 * 1e6

print(o.shape)

print(f"Time taken for Bessel function (slow): {b_time_slow:.6f} microseconds vs matlab {mat_b_time:.6f} microseconds")
print(f"Time taken for Bessel function (fast): {b_time_fast:.6f} microseconds vs matlab {mat_b_time:.6f} microseconds")
print(f"Time taken for Cosine function: {c_time:.6f} microseconds vs matlab {mat_c_time:.6f} microseconds")
print(f"Bessel to cosine ratio: {b_time_slow/c_time:.2f} vs matlab {mat_b_time/mat_c_time:.2f}")
print(f"Bessel to cosine ratio (fast): {b_time_fast/c_time:.2f} vs matlab {mat_b_time/mat_c_time:.2f}")
```
